```python
from astropy.io import fits
from astropy.stats import sigma_clipped_stats
```

```python
data = fits.getdata("E:/Documents/RGC Project/Data/vla-set-1/vla_first_fits_1/J065544.3+631921.fits")
```

```python
mean, median, std = sigma_clipped_stats(data, sigma=3.0)
print("Background RMS (noise):", std)
```

```python
from photutils.background import Background2D, MedianBackground

bkg = Background2D(data, box_size=50, filter_size=3,
                   bkg_estimator=MedianBackground())

noise = bkg.background_rms_median
print("Background RMS:", noise)
```
